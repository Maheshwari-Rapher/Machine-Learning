# Lasso - Ridge Regularization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# import Linear Regression ML libraries
from sklearn import preprocessing
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression , Ridge, Lasso
from sklearn.metrics import r2_score
```

```python
%matplotlib inline
```

```python
data =pd.read_csv(r"C:\Users\Maheshwari\Downloads\ML_NIT\Regression_nit\Regularization\car-mpg.csv")
data
```

```python
# Data cleansing 
data = data.drop(['car_name'],axis = 1)               #drop the car name
data['origin'] = data['origin'].replace({1:'america',2:'europe',3:'asia'})  #replace origin into 1,2,3, dont forgot get_dummies
data = pd.get_dummies(data,columns = ['origin'],dtype=int)     
data = data.replace('?',np.nan)          #replace ? to nan 
```

```python
data = data.apply(pd.to_numeric,errors='ignore')

#fill the missing values with median only for numeric columns
numeric_col = data.select_dtypes(include=[np.number]).columns
data[numeric_col] = data[numeric_col].apply(lambda x: x.fillna(x.median()))
```

```python
data.head()
```

 we have to predict the mpg column given the features

# 2 . Model Building

```python
x = data.drop(['mpg'], axis = 1)      # independent variable
y = data[['mpg']]                     # dependent variable
```

```python
# Scalaing the data
x_s = preprocessing.scale(x)
x_s = pd.DataFrame(x_s,columns = x.columns)  # converting scaled data into dataframe

y_s =preprocessing.scale(y)
y_s = pd.DataFrame(y_s,columns = y.columns) # ideally train,test data should be in columns
```

```python
x_s
```

```python
y_s
```

```python
data.shape
```

```python
# split into train  and test set
x_train,x_test,y_train,y_test = train_test_split(x_s, y_s, test_size = 0.20 ,random_state = 0)
```

```python
x_train.shape            
```

## 2(a). simple Linear Model

```python
# build the simple Linear model and find coefficients
regression_model = LinearRegression()
regression_model.fit(x_train,y_train)

for idx,col_name in enumerate(x_train.columns):
    print('The coefficient for {} is {} '.format(col_name,regression_model.coef_[0][idx]))

intercept = regression_model.intercept_[0]
print('The intercept is {}'.format(intercept))
```

## 2(b). Regularized Ridge Regression

```python
# Alpha factor 
# here is the lambda (penalty term) which helps to reduce the magnitude of coef

ridge_model = Ridge(alpha = 0.4)            # random value(0.4)
ridge_model.fit(x_train,y_train)                #alpha = loss fn + penalty

print('Ridge model coef: {}'.format(ridge_model.coef_))
#data as 10 columns so 10 coefficients are appear here
```

## 2(c). Regularized Lasso Regression

```python
# Alpha factor 
# here is the lambda (penalty term) which helps to reduce the magnitude of coef

lasso_model = Lasso(alpha = 0.1)            # random value(0.1)
lasso_model.fit(x_train,y_train)                #alpha = loss fn + penalty

print('Lasso model coef: {}'.format(lasso_model.coef_))
#data as 10 columns so 10 coefficients are appear here
```

# 3 . Score  comparison

```python
# simple linear model
print(regression_model.score(x_train,y_train))
print(regression_model.score(x_test,y_test))

print('*****************************')



#Ridge
print(ridge_model.score(x_train,y_train))
print(ridge_model.score(x_test,y_test))

print('*******************************')



#Lasso
print(lasso_model.score(x_train,y_train))
print(lasso_model.score(x_test,y_test))


print('*********************************')
```
