## Encode ? as NaNs

```python
df[df == '?'] = np.nan
```

## Impute the missing values with mode strategy

```python
for col in ['workclass', 'occupation', 'native.country']:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

## Check again for missing valuesa

```python
df.isnull().sum()    
```

now there are no missing values in the dataset

## Divide the dataset into x and y

```python
X = df.drop(['income'], axis=1)

y = df['income']
```

```python
X.head()
```

```python
y.head()   # here income is the dependent variable
```

## Split the data into train and test sets

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 0)
```

# Feature Engineering ( EDA )

## Encode categorical variables

```python
from sklearn import preprocessing

categorical = ['workclass', 'education', 'marital.status', 'occupation', 'relationship', 'race', 'sex', 'native.country']
for feature in categorical:
        le = preprocessing.LabelEncoder()
        X_train[feature] = le.fit_transform(X_train[feature])
        X_test[feature] = le.transform(X_test[feature])
```

## Feature scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train = pd.DataFrame(scaler.fit_transform(X_train),columns = X.columns)
X_test = pd.DataFrame(scaler.transform(X_test),columns = X.columns)
```

```python
X_train.head()
```

- all the dataset is converted to -3 to +3

## Logistic Regression model with all features

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

logreg = LogisticRegression()
logreg.fit(X_train,y_train)
y_pred = logreg.predict(X_test)
print('Logistic Regression accuracy score with all the features:{0:0.4f}'.format(accuracy_score(y_test,y_pred)))
```

## Logistic Regression with PCA

```python
from sklearn.decomposition import PCA
pca = PCA()
X_train = pca.fit_transform(X_train)
pca.explained_variance_ratio_
```

## Logistic Regression  with first 13 attributes

```python
X = df.drop(['income','native.country'], axis=1)
y = df['income']


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 0)


categorical = ['workclass', 'education', 'marital.status', 'occupation', 'relationship', 'race', 'sex']
for feature in categorical:
        le = preprocessing.LabelEncoder()
        X_train[feature] = le.fit_transform(X_train[feature])
        X_test[feature] = le.transform(X_test[feature])


X_train = pd.DataFrame(scaler.fit_transform(X_train), columns = X.columns)

X_test = pd.DataFrame(scaler.transform(X_test), columns = X.columns)

logreg = LogisticRegression()
logreg.fit(X_train, y_train)
y_pred = logreg.predict(X_test)

print('Logistic Regression accuracy score with the first 13 features: {0:0.4f}'. format(accuracy_score(y_test, y_pred)))
```
